```python
x = 56
y = int(input("Aklımdan bir sayı tuttum hadi bil! "))
y = ((y*y) - y)

if(y==x):
  print("Tebrikler 100 puan kazandınız!")
elif(y != x):
  result = abs(x - y)
  if(result <= 10):
    print("Tebrikler 50 puan kazandınız!")
  elif(11 <= result <= 50):
    print("Yaklaştın yeniden dene!")
  elif(result > 50):
    print("Kaybettiniz")
```
